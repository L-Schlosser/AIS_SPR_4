# Download the Dataset

*Source:* https://huggingface.co/datasets/Aoschu/German_invoices_dataset?library=datasets

```python
from datasets import load_dataset
import os

# Load the dataset
ds = load_dataset("Aoschu/German_invoices_dataset")

# Save all images to a folder
os.makedirs("invoice_images", exist_ok=True)

for i, row in enumerate(ds["train"]):
    img = row["image"]          # This is a PIL Image object
    img.save(f"invoice_images/invoice_{i:03d}.jpg")

print(f"Saved {len(ds['train'])} images.")
```
