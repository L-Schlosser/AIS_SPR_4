## Doctors Notes

```python
# Importing necessary library
import Augmentor
# Passing the path of the image directory
p = Augmentor.Pipeline(r'C:\Users\Celina Binder\Documents\FH_Hagenberg\Semester_Project\AIS_SPR_4\data\TestData\DoctorsNote')

# Defining augmentation parameters and generating 5 samples
p.flip_left_right(0.5)
p.black_and_white(0.1)
p.rotate(0.3, 10, 10)
p.skew(0.4, 0.1)
p.zoom(probability = 0.2, min_factor = 0.6, max_factor = 1.1)
p.sample(5)
```
