```python
import os
import numpy as np
from pdf2image import convert_from_path
from PIL import Image, ImageFilter, ImageOps
import random

def augment(img):
    # Grayscale
    if random.random() < 0.1:
        img = ImageOps.grayscale(img).convert("RGB")
    # Rotate
    img = img.rotate(random.uniform(-5, 10), expand=True, fillcolor=(255,255,255))
    # Shear via transform
    if random.random() < 0.3:
        shear = random.uniform(-0.1, 0.1)
        img = img.transform(img.size, Image.AFFINE, (1, shear, 0, 0, 1, 0))
    # Zoom/scale
    if random.random() < 0.2:
        scale = random.uniform(0.6, 1.1)
        new_size = (int(img.width * scale), int(img.height * scale))
        img = img.resize(new_size, Image.LANCZOS)
    # Blur as proxy for distortion
    if random.random() < 0.3:
        img = img.filter(ImageFilter.GaussianBlur(radius=random.uniform(0.5, 1.5)))
    return img

PDF_DIR    = r'C:\Users\Celina Binder\Documents\FH_Hagenberg\Semester_Project\AIS_SPR_4\data\TestData\DoctorsNote'
OUTPUT_DIR = os.path.join(PDF_DIR, 'output')
NUM_SAMPLES = 15
os.makedirs(OUTPUT_DIR, exist_ok=True)

for filename in os.listdir(PDF_DIR):
    if filename.lower().endswith('.pdf'):
        pages = convert_from_path(os.path.join(PDF_DIR, filename), dpi=200, first_page=1, last_page=1)
        for page in pages:
            for i in range(NUM_SAMPLES):
                augment(page).save(os.path.join(OUTPUT_DIR, f"{filename[:-4]}_aug{i+1}.png"))

print("Done!")
```
